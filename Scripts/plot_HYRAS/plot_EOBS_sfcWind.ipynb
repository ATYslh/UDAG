```python
import matplotlib.lines as mlines
import matplotlib.pyplot as plt


def load_json_file(path: str) -> dict:
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)


def mean_abs_diff(EOBS_path, nukleus_hereon, nukleus_btu, udag_eval, var_name, freq):

    with xr.open_dataset(EOBS_path) as dx:
        dx = dx.sel(time=slice("2005", "2014"))
        with xr.open_dataset(nukleus_hereon) as df:
            df = df.sel(time=slice("2005", "2014"))
            diff = (np.squeeze(df[var_name].values) - np.squeeze(dx[var_name].values),)
            print(
                f"Mean abs. diff Nukleus Hereon {variable} {freq}: {np.mean(np.abs(diff)):.2f}"
            )

        with xr.open_dataset(nukleus_btu) as df:
            df = df.sel(time=slice("2005", "2014"))
            diff = (np.squeeze(df[var_name].values) - np.squeeze(dx[var_name].values),)
            print(
                f"Mean abs. diff Nukleus BTU {variable} {freq}: {np.mean(np.abs(diff)):.2f}"
            )

        with xr.open_dataset(udag_eval) as df:
            df = df.sel(time=slice("2005", "2014"))
            diff = np.squeeze(df[list(df.keys())[1]].values) - np.squeeze(
                dx[var_name].values
            )
            print(f"Mean abs. diff UDAG {variable} {freq}: {np.mean(np.abs(diff)):.2f}")


def moving_average(x, w):
    return np.convolve(x, np.ones(w), "valid") / w


def three_day_mean_abs_diff(
    EOBS_path, nukleus_hereon, nukleus_btu, udag_eval, var_name, freq
):
    freq = "3 day mean"
    with xr.open_dataset(EOBS_path) as dx:
        dx = dx.sel(time=slice("2005", "2014"))
        with xr.open_dataset(nukleus_hereon) as df:
            df = df.sel(time=slice("2005", "2014"))
            diff = (
                moving_average(np.squeeze(df[var_name].values), 3)
                - moving_average(np.squeeze(dx[var_name].values), 3),
            )
            print(
                f"Moving mean abs. diff Nukleus Hereon {variable} {freq}: {np.mean(np.abs(diff)):.2f}"
            )

        with xr.open_dataset(nukleus_btu) as df:
            df = df.sel(time=slice("2005", "2014"))
            diff = (
                moving_average(np.squeeze(df[var_name].values), 3)
                - moving_average(np.squeeze(dx[var_name].values), 3),
            )
            print(
                f"Moving mean abs. diff Nukleus BTU {variable} {freq}: {np.mean(np.abs(diff)):.2f}"
            )

        with xr.open_dataset(udag_eval) as df:
            df = df.sel(time=slice("2005", "2014"))
            diff = moving_average(
                np.squeeze(df[list(df.keys())[1]].values), 3
            ) - moving_average(np.squeeze(dx[var_name].values), 3)
            print(
                f"Moving mean abs. diff UDAG {variable} {freq}: {np.mean(np.abs(diff)):.2f}"
            )


base_path = "/work/bb1364/g260190_heinrich/UDAG/Data/json_files"

files = {
    "NUKLEUS": {
        "sfcWind": load_json_file(f"{base_path}/NUKLEUS_Germany_sfcWind_info.json")
    },
    "EOBS": {"sfcWind": load_json_file(f"{base_path}/EOBS_Germany_sfcWind_info.json")},
    "UDAG": {"sfcWind": load_json_file(f"{base_path}/UDAG_Germany_sfcWind_info.json")},
}

for variable in ["sfcWind"]:
    for freq in ["yearly", "mon", "day"]:
        mean_abs_diff(
            files["EOBS"][variable][freq],
            files["NUKLEUS"][variable][freq]["ERA5"]["evaluation"]["CEU-3"][1],
            files["NUKLEUS"][variable][freq]["ERA5"]["evaluation"]["CEU-3"][0],
            files["UDAG"][variable][freq]["ERA5"]["evaluation"]["MEU-3"][0],
            variable,
            freq,
        )

    three_day_mean_abs_diff(
        files["EOBS"][variable][freq],
        files["NUKLEUS"][variable][freq]["ERA5"]["evaluation"]["CEU-3"][1],
        files["NUKLEUS"][variable][freq]["ERA5"]["evaluation"]["CEU-3"][0],
        files["UDAG"][variable][freq]["ERA5"]["evaluation"]["MEU-3"][0],
        variable,
        "day",
    )
```

```python
import matplotlib.lines as mlines
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# Helpers
# ---------------------------------------------------------


def load_json_file(path: str) -> dict:
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)


def monthly_mean(ds, var):
    return ds[var].groupby("time.month").mean()


# ---------------------------------------------------------
# File loading
# ---------------------------------------------------------

base_path = "/work/bb1364/g260190_heinrich/UDAG/Data/json_files"

files = {
    "NUKLEUS": {
        "sfcWind": load_json_file(f"{base_path}/NUKLEUS_Germany_sfcWind_info.json"),
    },
    "EOBS": {
        "sfcWind": load_json_file(f"{base_path}/EOBS_Germany_sfcWind_info.json"),
    },
    "UDAG": {
        "sfcWind": load_json_file(f"{base_path}/UDAG_Germany_sfcWind_info.json"),
    },
}

freq = "mon"
var_name = "sfcWind"

# ---------------------------------------------------------
# Dataset definitions
# ---------------------------------------------------------

datasets = {
    "NUKLEUS CCLM Hereon": files["NUKLEUS"][var_name][freq]["ERA5"]["evaluation"][
        "CEU-3"
    ][1],
    "NUKLEUS ICON BTU": files["NUKLEUS"][var_name][freq]["ERA5"]["evaluation"]["CEU-3"][
        0
    ],
    "UDAG ICON": files["UDAG"][var_name][freq]["ERA5"]["evaluation"]["MEU-3"][0],
}

colors = {
    "EOBS": "blue",
    "NUKLEUS CCLM Hereon": "orange",
    "NUKLEUS ICON BTU": "green",
    "UDAG ICON": "red",
}

# ---------------------------------------------------------
# Load EOBS
# ---------------------------------------------------------

with xr.open_dataset(files["EOBS"][var_name][freq]) as ds_eobs:
    ds_eobs = ds_eobs.sel(time=slice("2005", "2015"))
    eobs_monthly = monthly_mean(ds_eobs, var_name)

# ---------------------------------------------------------
# Load all model datasets and compute monthly means + differences
# ---------------------------------------------------------

monthly_means = {}
monthly_diffs = {}

for name, path in datasets.items():
    with xr.open_dataset(path) as ds:
        ds = ds.sel(time=slice("2005", "2015"))
        monthly_means[name] = monthly_mean(ds, var_name)

        # index-based difference to EOBS
        diff = xr.DataArray(
            monthly_means[name].values - eobs_monthly.values,
            dims=eobs_monthly.dims,
            coords=eobs_monthly.coords,
        )
        monthly_diffs[name] = diff

# ---------------------------------------------------------
# Compute global min/max across all curves and bars
# ---------------------------------------------------------

all_monthly = [eobs_monthly] + list(monthly_means.values())
all_diffs = list(monthly_diffs.values())

min_value = xr.concat(all_monthly + all_diffs, dim="source").min().item()
max_value = xr.concat(all_monthly + all_diffs, dim="source").max().item()

# ---------------------------------------------------------
# Plotting
# ---------------------------------------------------------

fig, ax = plt.subplots(figsize=(14, 9))
months = np.arange(1, 13)
# EOBS curve
ax.plot(months, eobs_monthly, color=colors["EOBS"], label="EOBS")


# Model curves
for name, monthly in monthly_means.items():
    ax.plot(months, monthly, color=colors[name], label=name)

ax.set_ylabel("sfcWind [m/s]")
ax.set_title("Germany sfcWind - Annual cycle")
ax.set_ylim(min_value * 1.02, max_value * 1.02)

# Side-by-side bars for differences
ax2 = ax.twinx()

bw = 0.25  # bar width

for i, (name, diff) in enumerate(monthly_diffs.items()):
    offset = (i - 1) * bw  # centers bars around each month
    ax2.bar(months + offset, diff.values, width=bw, alpha=0.4, color=colors[name])


ax2.set_ylim(min_value * 1.02, max_value * 1.02)
ax2.set_ylabel("Difference to Eobs [m/s]")

# Legend
handles = [mlines.Line2D([], [], color=colors[k], label=k) for k in colors]
ax.legend(handles=handles, loc="lower center", bbox_to_anchor=(0.5, -0.15), ncol=4)
ax.set_xlim(0.5, 12.5)
month_names = [
    "Jan",
    "Feb",
    "Mar",
    "Apr",
    "May",
    "Jun",
    "Jul",
    "Aug",
    "Sep",
    "Oct",
    "Nov",
    "Dec",
]

ax.set_xticks(np.arange(1, 13))
ax.set_xticklabels(month_names)

plt.tight_layout()
plt.savefig(
    f"/work/bb1364/g260190_heinrich/UDAG/Visualisation/EVAL_EOBS_sfcWind_AC_{freq}.png",
    bbox_inches="tight",
    format="png",
    dpi=300,
)
plt.show()
```

```python
import matplotlib.lines as mlines
import matplotlib.pyplot as plt
import numpy as np
import xarray as xr

# ---------------------------------------------------------
# Helpers
# ---------------------------------------------------------


def load_json_file(path: str) -> dict:
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)


def monthly_mean(ds, var):
    return ds[var].groupby("time.month").mean()


# ---------------------------------------------------------
# File loading
# ---------------------------------------------------------

base_path = "/work/bb1364/g260190_heinrich/UDAG/Data/json_files"

files = {
    "NUKLEUS": {
        "sfcWind": load_json_file(f"{base_path}/NUKLEUS_Germany_sfcWind_info.json"),
    },
    "EOBS": {
        "sfcWind": load_json_file(f"{base_path}/EOBS_Germany_sfcWind_info.json"),
    },
    "UDAG": {
        "sfcWind": load_json_file(f"{base_path}/UDAG_Germany_sfcWind_info.json"),
    },
}

freq = "mon"
var_name = "sfcWind"

# ---------------------------------------------------------
# Dataset definitions
# ---------------------------------------------------------

datasets = {
    "NUKLEUS CCLM Hereon": files["NUKLEUS"][var_name][freq]["ERA5"]["evaluation"][
        "CEU-3"
    ][1],
    "NUKLEUS ICON BTU": files["NUKLEUS"][var_name][freq]["ERA5"]["evaluation"]["CEU-3"][
        0
    ],
    "UDAG ICON": files["UDAG"][var_name][freq]["ERA5"]["evaluation"]["MEU-3"][0],
}

colors = {
    "EOBS": "blue",
    "NUKLEUS CCLM Hereon": "orange",
    "NUKLEUS ICON BTU": "green",
    "UDAG ICON": "red",
}

# ---------------------------------------------------------
# Time windows per dataset
# ---------------------------------------------------------

time_windows = {
    "NUKLEUS CCLM Hereon": ("2005", "2015"),
    "NUKLEUS ICON BTU": ("2005", "2014"),  # special case
    "UDAG ICON": ("2005", "2015"),
}

# ---------------------------------------------------------
# Load EOBS full (2005–2015) for plotting the EOBS curve
# ---------------------------------------------------------

with xr.open_dataset(files["EOBS"][var_name][freq]) as ds_eobs_full:
    ds_eobs_full = ds_eobs_full.sel(time=slice("2005", "2015"))
    eobs_monthly_full = monthly_mean(ds_eobs_full, var_name)

# ---------------------------------------------------------
# Load all model datasets and compute monthly means + differences
# ---------------------------------------------------------

monthly_means = {}
monthly_diffs = {}

for name, path in datasets.items():
    start, end = time_windows[name]

    # Load model dataset with its own time window
    with xr.open_dataset(path) as ds:
        ds = ds.sel(time=slice(start, end))
        monthly_means[name] = monthly_mean(ds, var_name)

    # Load matching EOBS window for this dataset
    with xr.open_dataset(files["EOBS"][var_name][freq]) as ds_eobs_local:
        ds_eobs_local = ds_eobs_local.sel(time=slice(start, end))
        eobs_monthly_local = monthly_mean(ds_eobs_local, var_name)

    # Compute difference (model - EOBS) for the matching window
    diff = xr.DataArray(
        monthly_means[name].values - eobs_monthly_local.values,
        dims=eobs_monthly_local.dims,
        coords=eobs_monthly_local.coords,
    )
    monthly_diffs[name] = diff

# ---------------------------------------------------------
# Compute global min/max across all curves and bars
# ---------------------------------------------------------

all_monthly = [eobs_monthly_full] + list(monthly_means.values())
all_diffs = list(monthly_diffs.values())

min_value = xr.concat(all_monthly + all_diffs, dim="source").min().item()
max_value = xr.concat(all_monthly + all_diffs, dim="source").max().item()

# ---------------------------------------------------------
# Plotting
# ---------------------------------------------------------

fig, ax = plt.subplots(figsize=(14, 9))
months = np.arange(1, 13)

# EOBS curve (always 2005–2015)
ax.plot(months, eobs_monthly_full, color=colors["EOBS"], label="EOBS")

# Model curves
for name, monthly in monthly_means.items():
    ax.plot(months, monthly, color=colors[name], label=name)

ax.set_ylabel("sfcWind [m/s]")
ax.set_title("Germany sfcWind – Annual cycle")
ax.set_ylim(min_value * 1.02, max_value * 1.02)

# Side-by-side bars for differences
ax2 = ax.twinx()
bw = 0.25  # bar width

for i, (name, diff) in enumerate(monthly_diffs.items()):
    offset = (i - 1) * bw
    ax2.bar(months + offset, diff.values, width=bw, alpha=0.4, color=colors[name])

ax2.set_ylim(min_value * 1.02, max_value * 1.02)
ax2.set_ylabel("Difference to EOBS [m/s]")

# Legend
handles = [mlines.Line2D([], [], color=colors[k], label=k) for k in colors]
ax.legend(handles=handles, loc="lower center", bbox_to_anchor=(0.5, -0.15), ncol=4)

# Month labels
month_names = [
    "Jan",
    "Feb",
    "Mar",
    "Apr",
    "May",
    "Jun",
    "Jul",
    "Aug",
    "Sep",
    "Oct",
    "Nov",
    "Dec",
]
ax.set_xticks(np.arange(1, 13))
ax.set_xticklabels(month_names)
ax.set_xlim(0.5, 12.5)

plt.tight_layout()
plt.savefig(
    f"/work/bb1364/g260190_heinrich/UDAG/Visualisation/EVAL_EOBS_sfcWind_AC_{freq}.png",
    bbox_inches="tight",
    format="png",
    dpi=300,
)
plt.show()
```
